```python
import csv
import random

# הגדרות הקובץ
filename = 'group03_http_input.csv'
headers = ['msg_id', 'app_protocol', 'src_app', 'dst_app', 'message', 'timestamp']

# רשימה של דפים וקבצים לסימולציה של גלישה
pages = ['index.html', 'about.html', 'contact.html', 'products.html', 'login.html', 'style.css', 'script.js', 'logo.png', 'banner.jpg']
methods = ['GET', 'POST']
responses = ['200 OK', '404 Not Found', '500 Internal Server Error', '301 Moved Permanently']

data = []
current_time = 0.0

# יצירת 65 אינטראקציות (כל אינטראקציה היא בקשה + תשובה = 130 שורות סה"כ)
for i in range(1, 66):
    # --- הבקשה (Request) ---
    msg_id_req = (i * 2) - 1
    page = random.choice(pages)
    method = 'POST' if page == 'login.html' else 'GET'
    
    request_msg = f"{method} /{page} HTTP/1.1"
    
    # קידום זמן אקראי קטן
    current_time += round(random.uniform(0.01, 0.5), 3)
    
    data.append([msg_id_req, 'HTTP', 'client_browser', 'web_server', request_msg, f"{current_time:.3f}"])

    # --- התשובה (Response) ---
    msg_id_res = i * 2
    
    # רוב הזמן התשובה תקינה, לפעמים שגיאה
    if random.random() > 0.9:
        response_status = random.choice(responses[1:]) # שגיאה אקראית
    else:
        response_status = responses[0] # 200 OK
        
    response_msg = f"HTTP/1.1 {response_status}"
    
    # זמן עיבוד קצר בשרת
    current_time += round(random.uniform(0.005, 0.1), 3)
    
    data.append([msg_id_res, 'HTTP', 'web_server', 'client_browser', response_msg, f"{current_time:.3f}"])

# כתיבה לקובץ CSV
with open(filename, 'w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(headers)
    writer.writerows(data)

print(f"File '{filename}' created successfully with {len(data)} lines of data.")
```
